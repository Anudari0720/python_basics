:Variable = A container for a value (string, integer, float, boolean)
A variable behaves as if was the value it contains


```python
#1 Strings-(Мөр) гэж юу вэ?
# Хашилт дотор (" ") байгаа бүх текстийг String гэнэ.
# "17" esvel "94567899" gesen toonud ch hashilttai bol tekst bolno.
first_name = "Anu"
food = "pizza"
age = "17"
email = "Darianu12@gamil.com"
phone_num = "94567899"
#2 f" " ashiglan tekst dund huvisagchiig {} haaltand hiij hevledeg.
print(f"Hello {first_name}")
print(f"You like {food}")
print(f"Your age {age}")
print(f"Your email {email}")
print(f"Your phone number is {phone_num}")
```

```python
 # integers-(БҮХЭЛ ТОО) ГЭЖ ЮУ ВЭ?
 # Хашилтгүй бичигдсэн, таслалгүй бүх эерэг, сөрөг, тэг тоог хэлнэ.
 # Компьютер эдгээр тоон дээр жинхэнэ математик үйлдэл хийж чадна.

age = 17
quantity = 4
num_of_students = 37
num_of_family = 3
grade_of_school = 12

# f-string ашиглан хэвлэх хэсэг:
print(f"You are {age} years old")
print(f"You are buying {quantity} items")
print(f"Your class has {num_of_students} students")
print(f"Your family has {num_of_family} people")
print(f"your grade has {grade_of_school} school")
```

```python
# Float (БУТАРХАЙ ТОО) ГЭЖ ЮУ ВЭ?
# Цэгтэй бичигдсэн бүх бутархай тоонуудыг хэлнэ.
# Компьютер үүнийг Integer (Бүхэл тоо)-оос өөрөөр хардаг.

num = 12.8
price = 99.9
gpa = 4.5
distance = 2.4

# f-string ашиглан хэвлэх хэсэг:
print(f"The number is {num}")
print(f"The price is {price}$")
print(f"My gpa is: {gpa}")
print(f"I ran {distance}km")
```

```python
# Boolean (ЛОГИК УТГА) ГЭЖ ЮУ ВЭ?
# Зөвхөн True (Үнэн) эсвэл False (Худал) гэсэн хоёрхон утга авдаг.
# Хашилтгүй, эхний үсгийг нь заавал ТОМООР бичнэ.

is_student = False
if is_student:
  print("You are a student")
else:
  print("You aren't a student")

is_online = True
if is_online:
  print("You are online")
else:
  print("You are offline")

```
